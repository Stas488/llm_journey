```python
import numpy as np
```

```python
a = [2, 4, -1]
b = [1, -2, 3]
```

```python
res = 0
for elem1, elem2 in zip(a,b):
    res+=elem1*elem2

res_dot = np.dot(a,b)

print(f'Результат через for {res:.5f}\nРезультат через np.dot {res_dot:.5f}')
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Наши векторы
v1 = np.array([3, 1])
v2 = np.array([1, 3])

plt.figure(figsize=(6, 6))

# Рисуем стрелки (quiver)
# quiver(X, Y, U, V) где X,Y - начало стрелки, U,V - куда она смотрит
plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue', label='v1 [3, 1]')
plt.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='red', label='v2 [1, 3]')

# Настраиваем оси, чтобы они проходили через 0
plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

plt.grid(True)
plt.legend()
plt.title("Визуализация 2D-векторов")
plt.show()
```

```python
x = 10.0
lr = 0.1

for step in range(50):
    # 1. Считаем градиент (производную) в текущей точке
    grad = 2 * x - 4
    
    # 2. Обновляем x, двигаясь ПРОТИВ градиента
    x = x - lr * grad
    
    print(f"Шаг {step+1}: x = {x:.4f}, f(x) = {x**2 - 4*x + 5:.4f}")
```
